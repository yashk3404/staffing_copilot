# Day 4 (Part 2): Skill Taxonomy Deep Dive
Validate and explore the 45-skill taxonomy before building the data generator

```python
import pandas as pd
import matplotlib.pyplot as plt

skills_df = pd.read_csv("../data/processed/skills_taxonomy.csv")

# How many skills per category?
cat_counts = skills_df["category"].value_counts()

fig, ax = plt.subplots(figsize=(10, 5))
cat_counts.plot(kind="bar", ax=ax, color="steelblue", edgecolor="white")
ax.set_title("Skills per Category in Taxonomy", fontsize=13, fontweight="bold")
ax.set_xlabel("Category"); ax.set_ylabel("Number of Skills")
ax.tick_params(axis="x", rotation=45)
plt.tight_layout()
plt.savefig("../data/processed/skills_by_category.png", dpi=150)
plt.show()
print("✅ Saved chart")
```

```python
# Which roles appear most across skills? (role coverage)
from collections import Counter

all_roles = []
for roles_str in skills_df["related_roles"]:
    all_roles.extend([r.strip() for r in roles_str.split(";")])

role_counts = Counter(all_roles)
print("Role coverage in taxonomy (how many skills mention each role):")
for role, count in role_counts.most_common():
    bar = "█" * count
    print(f"  {role:<25} {bar} ({count})")
```

```python
# Skills that apply to ALL roles (universal skills)
universal = skills_df[skills_df["related_roles"].str.contains("All")]
print("Universal skills (apply to all roles):")
print(universal[["skill_name", "category"]].to_string(index=False))

# ML/AI cluster — what we'll embed most heavily
ml_skills = skills_df[skills_df["category"] == "ML/AI"]
print("\nML/AI Skills:")
print(ml_skills[["skill_name", "related_roles"]].to_string(index=False))
```

## ✅ Taxonomy validated:
- 45 skills across 9 categories
- Backend Dev and Data Engineer have the most skill coverage
- Universal skills: Git, Agile/Scrum
- ML/AI cluster is rich enough for semantic differentiation
