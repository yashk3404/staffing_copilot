# Day 4: Scope Definition + Schema Design

## Project Scope
**Staffing Copilot** — an AI system that:
1. Takes a project's required roles + skills
2. Semantically matches employees using embeddings
3. Solves the assignment problem with OR-Tools (no double-booking)
4. Explains every decision in plain English via RAG + LLM

## What counts as a "match"?
- Role alignment (title similarity)
- Skill overlap (semantic, not just keyword)
- Availability % >= project minimum
- Experience years >= project minimum
- No double-booking across simultaneous projects

```python
import pandas as pd

# Load and inspect all 3 schemas
emp_schema   = pd.read_csv("../data/processed/employee_schema.csv")
proj_schema  = pd.read_csv("../data/processed/project_schema.csv")
skills_df    = pd.read_csv("../data/processed/skills_taxonomy.csv")

print("=== EMPLOYEE SCHEMA ===")
print(emp_schema.to_string(index=False))

print("\n=== PROJECT SCHEMA ===")
print(proj_schema.to_string(index=False))
```

```python
print("=== SKILL TAXONOMY ===")
print(f"Total skills: {len(skills_df)}")
print(f"\nCategories:\n{skills_df['category'].value_counts().to_string()}")
print(f"\nSample rows:")
print(skills_df.head(10).to_string(index=False))
```

```python
# Skills per category — what we support
by_category = skills_df.groupby("category")["skill_name"].apply(list)
for cat, skills in by_category.items():
    print(f"\n{cat}:")
    print("  " + ", ".join(skills))
```

## ✅ Schema Decisions Made
- Skills are a **fixed taxonomy** (45 skills) — not free text
- Proficiency: `Beginner | Intermediate | Expert`
- Availability: integer 0–100 (percentage)
- Cost bands: `A` (junior) · `B` (mid) · `C` (senior)
- All IDs: E001–E??? for employees, P001–P??? for projects
