```python
# Day 3: Hybrid Data Exploration Notebook

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------
# Load Synthetic Data
# -----------------------------
employees = pd.read_csv("C:/Users/KUMAR/Desktop/staffing_copilot/data/processed/employees.csv")
projects = pd.read_csv("C:/Users/KUMAR/Desktop/staffing_copilot/data/processed/projects.csv")

print("Synthetic Employees:")
print(employees.head())

print("Synthetic Projects:")
print(projects.head())

# -----------------------------
# Load Kaggle HR Dataset
# -----------------------------
kaggle_employees = pd.read_csv("C:/Users/KUMAR/Desktop/staffing_copilot/data/raw/Employee.csv")

print("Kaggle Employees:")
print(kaggle_employees.head())

# -----------------------------
# Explore Synthetic Dataset
# -----------------------------
# Histogram of experience
plt.figure(figsize=(8,5))
sns.histplot(employees['experience_years'], bins=15, kde=True)
plt.title("Synthetic Employee Experience Distribution")
plt.show()

# Bar chart of roles
plt.figure(figsize=(10,5))
sns.countplot(y=employees['role'])
plt.title("Synthetic Employees per Role")
plt.show()

# -----------------------------
# Explore Kaggle Dataset
# -----------------------------
# Salary distribution
plt.figure(figsize=(8,5))
sns.histplot(kaggle_employees['Salary'], bins=20, kde=True)
plt.title("Kaggle Employee Salary Distribution")
plt.show()

# Attrition by Department
plt.figure(figsize=(10,5))
sns.countplot(x='Department', hue='Attrition', data=kaggle_employees)
plt.title("Attrition by Department (Kaggle)")
plt.xticks(rotation=45)
plt.show()

# Average salary by JobRole
avg_salary = kaggle_employees.groupby("JobRole")["Salary"].mean().sort_values()
plt.figure(figsize=(10,5))
sns.barplot(x=avg_salary.values, y=avg_salary.index)
plt.title("Average Salary by Job Role (Kaggle)")
plt.show()

# -----------------------------
# Hybrid Merge: Synthetic + Kaggle
# -----------------------------
merged = employees.merge(
    kaggle_employees[['JobRole','Department','Salary','Attrition']],
    left_on='role',
    right_on='JobRole',
    how='left'
)

print("Merged Hybrid Dataset:")
print(merged.head())

# Example analysis: average salary by synthetic role
avg_salary_hybrid = merged.groupby("role")["Salary"].mean()
print("Average Salary by Synthetic Role (Hybrid):")
print(avg_salary_hybrid)

# Visualization: Salary vs Role (Hybrid)
plt.figure(figsize=(10,5))
sns.barplot(x=avg_salary_hybrid.values, y=avg_salary_hybrid.index)
plt.title("Hybrid Dataset: Average Salary by Role")
plt.show()
```
