```python
import pandas as pd
import numpy as np
from faker import Faker
import matplotlib.pyplot as plt
import seaborn as sns

fake = Faker()

# Generate synthetic employee dataset
employees = []
roles = ["Data Scientist", "ML Engineer", "Backend Developer", "Frontend Developer", "Project Manager"]
skills = ["Python", "SQL", "TensorFlow", "React", "AWS", "Docker"]

for i in range(20):
    employees.append({
        "id": i+1,
        "name": fake.name(),
        "role": np.random.choice(roles),
        "skills": np.random.choice(skills, size=np.random.randint(1,4), replace=False).tolist(),
        "experience_years": np.random.randint(1, 11)
    })

df = pd.DataFrame(employees)

# Explore dataset
print(df.head())
print(df.info())
print(df.describe())

# Group by role
avg_exp = df.groupby("role")["experience_years"].mean()
print(avg_exp)

# Visualization
plt.figure(figsize=(8,5))
sns.barplot(x=avg_exp.index, y=avg_exp.values)
plt.xticks(rotation=45)
plt.title("Average Experience by Role")
plt.show()
```
